```python
import pandas as pd

# 1. CSV-Datei laden
df = pd.read_csv("./mieten_gesamt.csv")

# 2. NaN entfernen
df = df.dropna()

# 3. Nur relevante Jahre auswählen (2021–2025)
df = df[df["Jahr"] >= 2021]

# 4. Durchschnitt pro Stadt berechnen
durchschnitt_mieten = df.groupby("Stadt")["Mietpreis"].mean().reset_index()

# 5. Aufsteigend sortieren → günstigste zuerst
top5 = durchschnitt_mieten.sort_values("Mietpreis").head(5)

# 6. Ausgabe
print("Top 5 günstigste Städte (Ø Mietpreis 2021–2025):")
for i, row in top5.iterrows():
    print(f"{row['Stadt'].capitalize()}: {row['Mietpreis']:.2f} €/m²")
```
